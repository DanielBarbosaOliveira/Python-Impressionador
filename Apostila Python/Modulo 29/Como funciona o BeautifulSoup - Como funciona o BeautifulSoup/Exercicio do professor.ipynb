```python
import re
import requests
from bs4 import BeautifulSoup

link = "https://coinmarketcap.com/"
requisicao = requests.get(link)
site = BeautifulSoup(requisicao.text, "html.parser")
```

```python
from pprint import pprint

tbody = site.find("tbody")
linhas = tbody.find_all("tr")
moedas = {}
for linha in linhas:
    try:
        nome = linha.find(class_="eWrlhi").text
        codigo = linha.find(class_="coin-item-symbol").text
        valores = linha.find_all(string=re.compile(r"\$"))
        preco = valores[0]
        percentuais = linha.find_all(string=re.compile("%"))
        
        for indice, percentual in enumerate(percentuais):
            if "dJLZma" in percentual.parent["class"]:
                percentuais[indice] = "-"+ percentual
        
        variacao_1h = percentuais[0]
        variacao_24h = percentuais[1]
        variacao_7d = percentuais[2]
        
        market_cap = valores[1]
        volume = valores[2]
        dic = {
            "nome": nome,
            "codigo": codigo,
            "preco": preco,
            "market_cap": market_cap,
            "volume": volume,
            "variacao_1h": variacao_1h,
            "variacao_24h": variacao_24h,
            "variacao_7d": variacao_7d,
        }
        moedas[nome] = dic
    except AttributeError:
        continue

pprint(moedas)
```
