```python
import time
from selenium import webdriver
from bs4 import BeautifulSoup

navegador = webdriver.Chrome()

try:
    navegador.get("https://coinmarketcap.com/")
    for i in range(5):
        navegador.execute_script(f"window.scrollTo(0, {i} * 1000)")
        time.sleep(1)  # Aguarda 5 segundos para garantir que a página carregou completamente
    html = navegador.page_source
    site = BeautifulSoup(html, "html.parser")
except Exception as e:
    print(f"Erro ao acessar a página: {e}")
finally:
    navegador.close()
```

```python
moedas = {}

tabela_coins = site.find("tbody")
linhas = tabela_coins.find_all("tr")
for indice, linha in enumerate(linhas):
    nome_moeda = linha.find(class_="coin-item-name").text
    print(nome_moeda)
    preco_moeda = linha.find(class_="sc-631098c-0").text
    print(preco_moeda)
    variacao_1h = linha.find_all(class_="sc-d5c03ba0-0")[0].text
    print(variacao_1h)
    variacao_24h = linha.find_all(class_="sc-d5c03ba0-0")[1].text
    print(variacao_24h)
    variacao_7d = linha.find_all(class_="sc-d5c03ba0-0")[2].text
    print(variacao_7d)
    market_cap = linha.find_all(class_="BasePopover_base__T5yOf")[3].text
    print(market_cap)
    
    moedas[nome_moeda] = {"preco": preco_moeda, "variacao_1h": variacao_1h, "variacao_24h": variacao_24h, "variacao_7d": variacao_7d, "market_cap": market_cap}
```

```python
for moeda in moedas:
    print(f"Moeda: {moeda}")
    print(f"Preço: {moedas[moeda]['preco']}")
    print(f"Variação 1h: {moedas[moeda]['variacao_1h']}")
    print(f"Variação 24h: {moedas[moeda]['variacao_24h']}")
    print(f"Variação 7d: {moedas[moeda]['variacao_7d']}")
    print(f"Market Cap: {moedas[moeda]['market_cap']}")
    print()
```
