```python
from selenium import webdriver
from selenium.webdriver.common.by import By

navegador = webdriver.Chrome()
navegador.maximize_window()
```

```python
navegador.get("file:///Users/danielbarbosadeoliveira/cursos/Python-Impressionador/Apostila Python/Modulo 28/Trabalhando com iFrames - pagina.html")
```

```python
# Entrar no iframe
elemento_iframe = navegador.find_element(By.ID, "newsletter_iframe")
navegador.switch_to.frame(elemento_iframe)
navegador.find_element(By.ID, "nome").send_keys("Daniel")
navegador.find_element(By.ID, "email").send_keys("daniel@example.com")
```

```python
navegador.switch_to.default_content()  # Voltar para o conteúdo principal
navegador.find_element(By.ID, "campo_busca_principal").send_keys("Mensagem fora do iframe")
```

```python
lista_iframes = navegador.find_elements(By.TAG_NAME, "iframe")
print(len(lista_iframes))
```
