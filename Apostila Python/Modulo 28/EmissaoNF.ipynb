```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options

options = Options()

options.add_argument(r"C:\Users\santo\AppData\Local\Google\Chrome\User Data\Default")
options.add_argument("--start-maximized")

prefs = {
            "download.default_directory": r"C:\Temp\Downloads",  # Change path as needed
            "download.prompt_for_download": False,
            "profile.default_content_settings.popups": 0
        }

options.add_experimental_option("prefs",prefs)

navegador = webdriver.Chrome(options=options)
    
navegador.maximize_window()
```

```python
import os

caminho = os.getcwd()
arquivo = caminho + r'\login.html'
navegador.get(arquivo)
time.sleep(1)
```

```python
def fazer_login():
    navegador.find_element(By.XPATH, '/html/body/div/form/input[1]').send_keys("Daniel")
    navegador.find_element(By.XPATH, '/html/body/div/form/input[2]').send_keys('Senha')
    navegador.find_element(By.XPATH, '/html/body/div/form/button').click()
```

```python
def emitir_nf(linha):
    navegador.find_element(By.NAME, 'nome').send_keys(linha.Cliente)
    navegador.find_element(By.NAME, 'endereco').send_keys(linha.Endereço) 
    navegador.find_element(By.NAME, 'bairro').send_keys(linha.Bairro)
    navegador.find_element(By.NAME, 'municipio').send_keys(linha.Municipio)
    navegador.find_element(By.NAME, 'cep').send_keys(linha['CEP'])
    navegador.find_element(By.NAME, 'uf').send_keys(linha['UF'])
    navegador.find_element(By.NAME, 'cnpj').send_keys(linha['CPF/CNPJ'])
    navegador.find_element(By.NAME, 'inscricao').send_keys(linha['Inscricao Estadual'])
    navegador.find_element(By.NAME, 'descricao').send_keys(linha['Descrição'])
    navegador.find_element(By.NAME, 'quantidade').send_keys(linha.Quantidade)
    navegador.find_element(By.NAME, 'valor_unitario').send_keys(linha['Valor Unitario'])
    navegador.find_element(By.NAME, 'total').send_keys(linha['Valor Total'])
    navegador.find_element(By.CLASS_NAME, 'registerbtn').click()
    navegador.refresh()
```

```python
#fazer_login()
df_nfs = pd.read_excel('NotasEmitir.xlsx')

for indice, linha in df_nfs[:1].iterrows():
    emitir_nf(linha)
    #time.sleep(2)

navegador.quit()
```
