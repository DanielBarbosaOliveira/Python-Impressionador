```python
from selenium import webdriver
from selenium.webdriver.common.by import By

navegador = webdriver.Chrome()
navegador.maximize_window()
```

```python
import os

caminho = os.getcwd()
arquivo = caminho + r"\Pagina Hashtag.html"
navegador.get(arquivo)
```

#### Botão que abre outra janela

```python
navegador.find_element(By.XPATH, '/html/body/section[2]/div/div[6]/figure/a/img').click()
```

```python
navegador.find_element(By.NAME, 'firstname').send_keys("daniel")
navegador.find_element(By.NAME, 'email').send_keys("dandan@hotmail.com")
navegador.find_element(By.NAME, 'phone').send_keys("99088840")
```
