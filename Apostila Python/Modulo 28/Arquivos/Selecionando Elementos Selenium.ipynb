##### Ex1: Link do Whatsapp

````python
link_whatsapp = navegador.find_element(By.XPATH, "/html/body/footer/div/div[1]/div[2]/div/a[2]").get_attribute("href")
print(link_whatsapp)
````

### Ex2: Imagens dos cursos

````python
link_imagem = navegador.find_element(By.XPATH, "/html/body/section[2]/div/div[4]/figure/a").get_attribute('href')
print(link_imagem)
````

### Pegar todos os links das imagens

````python
lista_elementos = navegador.find_elements(By.TAG_NAME, 'figure')
for elemento in lista_elementos:
    try:
        link = elemento.find_element(By.TAG_NAME, 'a').get_attribute("href")
        print(link)
    except:
        continue
````
