```python
# criar o navegador
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.wait import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

navegador = webdriver.Chrome()
#navegador.maximize_window()
navegador.get("https://www.hashtagtreinamentos.com.br")
```

```python
# EC WebDriverWait

element = WebDriverWait(navegador, 20).until(EC.element_to_be_clickable((By.CLASS_NAME, 'gtm-popup__close')))
element.click()
```

```python
# Usando um loop
while len(navegador.find_elements(By.CLASS_NAME, 'gtm-popup__close')) < 1:
    time.sleep(1)

time.sleep(10)
navegador.find_element(By.CLASS_NAME, 'gtm-popup__close').click()
```

```python
# Na marra, usando time.sleep
time.sleep(20)
navegador.find_element(By.CLASS_NAME, 'gtm-popup__close').click()
```
