## Oferecendo desconto para cliente com base na última compra

Suponha que você está trabalhando para uma empresa que deseja rastrear a atividade do cliente. Uma métrica que eles estão interessados é o tempo que passou desde a última transação do cliente. Se for muito tempo, eles podem oferecer um desconto para o cliente. Crie um script Python que mostra quanto tempo se passou desde a última compra do cliente. Se faz mais de 30 dias, mostre uma mensagem oferecendo um desconto para o cliente.

```python
from datetime import datetime

ultima_compra = datetime(2026, 2, 15)
dia_atual = datetime.today()
dias_desde_ultima_compra = dia_atual - ultima_compra

print(f"Fazem {dias_desde_ultima_compra.days} dias desde a última compra.")

if dias_desde_ultima_compra.days > 30:
    print(f"Você ganhou um desconto de 10%, aproveite!")
else:
    print("Estamos sentindo sua falta, peça uma pizza.")
```

## Data e hora em diferentes fusos horários

Uma empresa tem escritórios em São Paulo, Nova York e Tóquio. Crie um script Python que mostra a data e hora atuais nesses três fusos horários. Exiba, também, se estes escritórios estão abertos ou fechados (9h às 17h).


```python
from datetime import datetime, timezone, timedelta
from zoneinfo import ZoneInfo, available_timezones

fuso_sp = ZoneInfo('America/Sao_Paulo')
fuso_nw = ZoneInfo('America/New_York')
fuso_tk = ZoneInfo('Asia/Tokyo')
horario_atual_sp = datetime.today().now(fuso_sp).time()
horario_atual_nw = datetime.today().now(fuso_nw).time()
horario_atual_tk = datetime.today().now(fuso_tk).time()


print(f"Horário em São Paulo {horario_atual_sp}")

if horario_atual_sp.hour >= 9 and horario_atual_sp.hour < 17:
    print("Escritório de São Paulo está aberto")


print(f"\nHorário de Nova York {horario_atual_nw}")

if horario_atual_nw.hour >= 9 and horario_atual_nw.hour < 17:
    print("Escritório de Nova York está aberto")
    

print(f"\nHorário de Tóqui {horario_atual_tk}")

if horario_atual_tk.hour >= 9 and horario_atual_tk.hour < 17:
    print("Escritório de Tóquio está aberto")
else:
    print("Escritório de Tóquio está fechado")
```
