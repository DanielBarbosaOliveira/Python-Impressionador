```python
import pyodbc

dados_conexao = ("Driver={SQLite3 ODBC Driver};Server=localhost;Database=chinook.db")

conexao = pyodbc.connect(dados_conexao)

cursor = conexao.cursor()
```

```python
cursor.execute("UPDATE customers SET Email='dan.dam@hotmail.com' WHERE Email='luisg@embraer.com.br'")

cursor.commit()

cursor.close()
conexao.close()
```
