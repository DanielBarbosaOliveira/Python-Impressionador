### Exemplo simples

```python
class Vendedor:

    def __init__(self, nome_vendedor):
        self.vendas = 0
        self.nome = nome_vendedor
        self.meta = 500
        self.bonus = 0

    def vendeu(self, quantidade_vendas):
        self.vendas = quantidade_vendas
        self.calcular_bonus()

    def calcular_bonus(self):
        if self.vendas > self.meta:
            self.bonus = 30
        else:
            self.bonus = 0
```

```python
vendedor1 = Vendedor('Daniel')
vendedor1.vendeu(5300)

vendedor2 = Vendedor('Lucas')
vendedor2.vendeu(1500)

print(vendedor1.nome, vendedor2.nome)
print(vendedor1.vendas, vendedor2.vendas)
print(vendedor1.bonus, vendedor2.bonus)
```

### Refatorando com classe

```python
class MeuRobo:

    def __init__(self, tempo_espera):
        self.tempo_espera = tempo_espera
        pyautogui.PAUSE = 1

    def abrir_programa(self, programa):
        pyautogui.PAUSE = 1
        pyautogui.hotkey("command", 'space')
        pyautogui.write(programa)
        pyautogui.press("enter")

    def entrar_primeiro_site(self, site):
        pyautogui.click(x=66, y=50)
        pyautogui.click(x=314, y=63)
        self.aguardar()
        self.escrever_e_enter(site)
        self.aguardar()

    def entrar_site(self, site):
        pyautogui.click(x=314, y=63)
        self.escrever_e_enter(site)
        self.aguardar()

    def escrever_e_enter(self, texto):
        pyperclip.copy(texto)
        pyautogui.hotkey("command", "v")
        pyautogui.press("enter")

    def aguardar(self):
        time.sleep(self.tempo_espera)

    def pesquisar_no_campo(self, texto):
        self.escrever_e_enter(texto)
        self.aguardar()

    def clicar(self, x, y, botao='left'):
        pyautogui.click(x, y,  button=botao)

    def pegar_posicao(self):
        for i in range(5):
            print(f"pegando posicao em {5 - i} segundos")
            time.sleep(1)

        print(pyautogui.position())

    def extrair_link(self, x, y):
        self.clicar(x, y, "right")
        for i in range(6):
            pyautogui.press("down")
        pyautogui.press("enter")
        texto = pyperclip.paste()
        print(texto)
```

```python
robo = MeuRobo(3)
robo.abrir_programa('chrome')
robo.entrar_primeiro_site('https://www.hashtagtreinamentos.com/blog')
robo.clicar(x=1351, y=168)
robo.pesquisar_no_campo('classe')
robo.clicar(x=431, y=841)
robo.aguardar()
robo.extrair_link(1112, 954)
```
