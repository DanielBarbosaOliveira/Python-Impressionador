## Calculando a idade

Um usuário fornece sua data de nascimento no formato "dd/mm/aaaa". Crie um script Python que calcula a idade do usuário.


```python
data_nascimento = input(f"Informe sua data de nascimento no formato (dd/mm/aaaa)").strip()
```

```python
from datetime import datetime

dia_atual = datetime.now()

data_nascimento_time = datetime.strptime(data_nascimento, '%d/%m/%Y')

idade = dia_atual.year - data_nascimento_time.year
mes_atual = dia_atual.month
mes_aniversario = data_nascimento_time.month

idade = int(idade)

if mes_atual > mes_aniversario:
    idade += 1
if mes_atual < mes_aniversario:
    idade -= 1


#if data_nascimento_time.month < datetime.month
print(f"Cliente tem {idade} anos")
```
