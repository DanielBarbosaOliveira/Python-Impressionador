## Exercício

Você é um analista de RH e tem os salários de todos os funcionários da sua empresa em um array NumPy. Seu trabalho é identificar quantos funcionários ganham acima da média. Use o seguinte array para sua análise: `salarios = np.array([3000, 2500, 3500, 4000, 2000, 4500, 3000, 3800, 4800])`.

```python
import numpy as np

salarios = np.array([3000, 2500, 3500, 4000, 2000, 4500, 3000, 3800, 4800])
```

```python
salarios_acima_media = salarios[salarios > np.mean(salarios)]
media_salarios = np.mean(salarios)
print("Salários acima da média:", salarios_acima_media)
print(f"Quantidade de salários acima da média: {len(salarios_acima_media)}")
```

```python
print(salarios > media_salarios)
```

```python
print(np.sum(salarios > media_salarios))  # Quantidade de salários acima da média
```

```python
print(sum([True, False, True, False]))  # Somar verdadeiros e falsos, True é 1 e False é 0
```

```python
funcionarios_acima_media = np.sum(salarios > media_salarios)
print(f"{funcionarios_acima_media} funcionários têm salários acima da média.")
```

## Bônus

Outras formas de fazer contagem

```python
print(np.count_nonzero(salarios > media_salarios))  # Outra forma de contar salários acima da média
```

```python
print(np.unique(salarios > media_salarios))  # Valores únicos de salários acima da média
```

```python
print(np.unique(salarios > media_salarios, return_counts=True))  # Valores únicos e suas contagens
```

```python
valores_unicos, contagens = np.unique(salarios > media_salarios, return_counts=True)
print("Valores únicos:", valores_unicos)
print("Contagens:", contagens)
```

## Bônus 2
Criando 20 valores fictícios de salários com a mesma faixa anterior, tomando cuidado de outras pessoas gerarem os mesmos valores:

```python
rng = np.random.default_rng(seed=42)  # Criando um gerador de números aleatórios com semente

salarios_gerados = rng.integers(low=np.min(salarios), high=np.max(salarios), size=20)
print(salarios_gerados)
```

```python
rng = np.random.default_rng(seed=42)  # Criando um gerador de números aleatórios com uma semente fixa

salarios_gerados = rng.integers(low=np.min(salarios), high=np.max(salarios), size=20, endpoint=True)  # Gerando 20 salários aleatórios entre o mínimo e máximo dos salários existentes, incluindo o máximo
print(salarios_gerados)
```

```python
print(np.mean(salarios_gerados))  # Média dos salários gerados
```

```python
print(salarios_gerados.min())
print(salarios_gerados.max())
print(salarios_gerados.mean())
```
