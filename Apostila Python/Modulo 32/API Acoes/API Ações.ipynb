##### Pegando cotações semanais - JSON

```python
import requests
from chave import chave_api
from pprint import pprint

url = f'https://www.alphavantage.co/query?function=TIME_SERIES_WEEKLY_ADJUSTED&symbol=ITUB3.SAO&apikey={chave_api}'

r = requests.get(url)
dic_requisicao = r.json()
print(dic_requisicao)
```

##### Pegando cotações semanais - CSV

```python
import pandas as pd
from io import StringIO

url = f'https://www.alphavantage.co/query?function=TIME_SERIES_WEEKLY_ADJUSTED&symbol=ITUB3.SAO&apikey={chave_api}&datatype=csv'

r = requests.get(url)

tabela = pd.read_csv(StringIO(r.text))
display(tabela)
```

#### Cotações

```python
acoes = ['ITUB4', 'ABEV3', 'BBAS3']

compilada = pd.DataFrame()

for acao in acoes:
    url = f'https://www.alphavantage.co/query?function=GLOBAL_QUOTE&symbol={acao}.SAO&apikey={chave_api}&datatype=csv'
    r = requests.get(url)
    tabela = pd.read_csv(StringIO(r.text))
    lista_tabelas = [compilada, tabela]
    compilada = pd.concat(lista_tabelas)

display(compilada)
```

### Encontrar ativos

```python
url = f'https://www.alphavantage.co/query?function=SYMBOL_SEARCH&keywords=amazon&apikey={chave_api}&datatype=csv'

r = requests.get(url)

tabela = pd.read_csv(StringIO(r.text))
display(tabela)
```
