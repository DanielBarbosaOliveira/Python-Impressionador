```python
from flask import Flask

app = Flask(__name__) # cria o site

@app.route("/") # decorator -> diz em qual link a função vai rodar
def helloworld():
    return {"Daniel": "brabo"}

@app.route("/danielmaroto") # decorator -> diz em qual link a função vai rodar
def daniel():
    return {"Daniel": "maroto"}

@app.route("/mentodaniel") # decorator -> diz em qual link a função vai rodar
def mentodaniel():
    return {"turma": "pythonimpressinador", "professor": "Daniel Barbosa de Oliveira", "nome": "mentodaniel"}

app.run() # roda o site
```
