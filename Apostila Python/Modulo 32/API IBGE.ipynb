### API IBGE
https://servicodados.ibge.gov.br/api/docs

```python
import requests
from pprint import pprint

link = "https://servicodados.ibge.gov.br/api/v3/agregados/7392/periodos/2014/variaveis/10484?localidades=N1[all]"

requisicao = requests.get(link)

informacoes = requisicao.json()
#pprint(informacoes[0])
informacoes = informacoes[0]
item_busca = informacoes["variavel"] # variavel
resultados = informacoes["resultados"][0]["series"] # series
```

```python
print(item_busca)
print(resultados)
```
