## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]

vendas_ano = vendas_1sem + vendas_2sem
melhor_mes = max(vendas_ano)
index_melhor_mes = vendas_ano.index(melhor_mes)

pior_mes = min(vendas_ano)
index_pior_mes = vendas_ano.index(pior_mes)print(f"Valor de vendas do pior mês: R$ {pior_mes:,.2f} {meses[index_pior_mes]}")
```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
print(f"O melhor mês do ano foi {meses[index_melhor_mes]} com vendas de R$ {melhor_mes:,.2f}")
print(f"O pior mês do ano foi {meses[index_pior_mes]} com vendas de R$ {pior_mes:,.2f}")

faturamento_total = sum(vendas_ano)
print(f"O faturamento total do ano foi de R$ {faturamento_total:,.2f}")
representacao_melhor_mes = (melhor_mes / faturamento_total)
print(f"A representação do melhor mês no faturamento total foi de {representacao_melhor_mes:.2%}%")
```
