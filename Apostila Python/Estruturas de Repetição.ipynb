#### 1. Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

````python
while True:
    nota = input("Digite uma nota entre 0 e 10: ")
    if not nota or not nota.isnumeric() or int(nota) > 10 or int(nota) < 0:
        print("Valor inválido. Tente novamente.")
    else:
        print(f"Nota: {nota}")
        break
    
````

#### 2. Faça um programa que leia um nome de usuário e a sua senha e não aceite a senha igual ao nome do usuário, mostrando uma mensagem de erro e voltando a pedir as informações.

````python
nome = input("Digite seu nome: ")
senha = input("Digite sua senha: ")
while nome == senha:
    print("O nome e a senha não podem ser iguais. Tente novamente.")
    nome = input("Digite seu nome: ")
    senha = input("Digite sua senha: ")
print("Cadastro realizado com sucesso!")
````

#### 3. Faça um programa que leia e valide as seguintes informações (e para cada uma delas, continue pedindo a informação até o usuário inserir corretamente):
##### Nome: maior que 3 caracteres;
##### Idade: entre 0 e 150;
##### Salário: maior que zero;
##### Sexo: 'f' ou 'm';
##### Estado Civil: 's', 'c', 'v', 'd';

````python
nome = input("Digite seu nome: ")
if not nome or len(nome) < 3 or nome.isnumeric():
    if not nome:
        while not nome:
            print("Digite o seu nome. Tente novamente.")
            nome = input("Digite seu nome: ")
    if len(nome) < 3:
        while len(nome) < 3:
            print("O nome deve ter pelo menos 3 caracteres. Tente novamente.")
            nome = input("Digite seu nome: ")
    if nome.isnumeric():
        while nome.isnumeric():
            print("O nome não pode ser um número. Tente novamente.")
            nome = input("Digite seu nome: ")

print("Nome cadastrado com sucesso!")

idade = input("Digite sua idade: ")
while not idade or not idade.isnumeric() or int(idade) < 0 or int(idade) > 150:
    if not idade:
        while not idade:
            print("Digite a sua idade. Tente novamente.")
            idade = input("Digite sua idade: ")
    if not idade.isnumeric():
        while not idade.isnumeric():
            print("A idade deve ser um número. Tente novamente.")
            idade = input("Digite sua idade: ")
    if int(idade) < 0:
        while int(idade) < 0:
            print("A idade não pode ser negativa. Tente novamente.")
            idade = input("Digite sua idade: ")
    if int(idade) > 150:
        while int(idade) > 150:
            print("A idade não pode ser maior que 150. Tente novamente.")
            idade = input("Digite sua idade: ")
print("Idade cadastrada com sucesso!")

salario = input("Digite seu salário:(Somente números) ")
while not salario or not salario.isnumeric() or int(salario) <= 0:
    if not salario:
        while not salario:
            print("Digite o seu salário. Tente novamente.")
            salario = input("Digite seu salário:(Somente números) ")
    if not salario.isnumeric():
        while not salario.isnumeric():
            print("O salário deve ser um número. Tente novamente.")
            salario = input("Digite seu salário:(Somente números) ")
    if int(salario) <= 0:
        while int(salario) <= 0:
            print("O salário deve ser um número positivo. Tente novamente.")
            salario = input("Digite seu salário:(Somente números) ")
print("Salário cadastrado com sucesso!")

sexo = input("Digite seu sexo (M/F): ").upper()
while sexo not in ["M", "F"]:
    print("Sexo inválido. Tente novamente.")
    sexo = input("Digite seu sexo (M/F): ").upper()
print("Sexo cadastrado com sucesso!")

estado_civil = input("Digite seu estado civil (S/C/V/D): ").upper()
while estado_civil not in ["S", "C", "V", "D"]:
    print("Estado civil inválido. Tente novamente.")
    estado_civil = input("Digite seu estado civil (S/C/V/D): ").upper()
````

#### 4. Supondo que a população de um país A seja da ordem de 80000 habitantes com uma taxa anual de crescimento de 3% e que a população de B seja 200000 habitantes com uma taxa de crescimento de 1.5%. Faça um programa que calcule e escreva o número de anos necessários para que a população do país A ultrapasse ou iguale a população do país B, mantidas as taxas de crescimento.

````python
paisa = 80000
paisb = 200000
anos = 0
while paisa < paisb:
    paisa += paisa * 0.03
    paisb += paisb * 0.015
    anos += 1
    print(f"Ano {anos}: País A - {paisa:.2f} habitantes, País B - {paisb:.2f} habitantes")
print(f"Serão necessários {anos} anos para que o país A ultrapasse o país B em população.")
````

#### 5. Altere o programa anterior permitindo ao usuário informar as populações e as taxas de crescimento iniciais. Valide a entrada e permita repetir a operação.
