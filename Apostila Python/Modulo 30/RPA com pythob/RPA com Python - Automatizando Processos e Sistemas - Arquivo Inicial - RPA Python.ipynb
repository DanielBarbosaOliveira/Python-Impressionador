## Desafio

- Vamos automatizar a extração de informações de um sistema e envio de um relatório por e-mail
- No nosso caso, para todo mundo conseguir fazer o mesmo programa, o nosso "sistema" vai ser o Gmail, mas o mesmo processo pode ser feito com qualquer programa do seu computador e qualquer sistema
    - Passo 1: Entrar no sistema (entrar no Gmail)
    - Passo 2: Entrar em uma aba específica do sistema onde tem o nosso relatório (Aba Contatos)
    - Passo 3: Exportar o Relatório (Exportar Contatos)
    - Passo 4: Pegar o relatório exportado, tratar e pegar as informações que queremos
    - Passo 5: Preencher/Atualizar informações do sistema (No nosso caso, criar um e-mail e enviar)

```python
import pyautogui
import time
```

```python
# pyautogui.write() -> escreve
# pyautogui.click() -> clica
# pyautogui.LocateOnScreen -> identifica uma imagem na tela
# pyautogui.hotkey() -> usa atalho do teclado
# pyautogui.press -> aperta um botão do teclado

pyautogui.alert("O código vai começar. Não mexa em NADA enquanto o código tiver rodando. Quando finalizar, eu te aviso.")

# abrir o sistema (abrir o navegador - Google Chrome)
pyautogui.PAUSE = 1
# apertar o windows do teclado
pyautogui.hotkey("command", "space")
# digitar Chrome
pyautogui.write("Chrome")
# apertar enter
pyautogui.press("enter")
# entrar no Gmail
gmail = pyautogui.locateCenterOnScreen("Gmail.png")
pyautogui.click(gmail)

# esperar o gmail
while True:
    try:
        pyautogui.locateOnScreen("escrever.png")
        time.sleep(1)
        break
    except:
        time.sleep(1)
# clicar no menu
menu = pyautogui.locateCenterOnScreen("menu.png")
pyautogui.click(menu)
# entrar em contatos
contatos = pyautogui.locateCenterOnScreen("contatos.png")
pyautogui.click(contatos)

# esprar o contatos
while True:
    try:
        pyautogui.locateOnScreen("contatos2.png")
        time.sleep(1)
        break
    except:
        time.sleep(1)

# exportar os contatos
exportar = pyautogui.locateCenterOnScreen("exportar.png")
pyautogui.click(exportar)
time.sleep(1)
exportar2 = pyautogui.locateCenterOnScreen("exportar2.png")
pyautogui.click(exportar2)
```

### Agora vamos escrever o e-mail

```python
import pandas as pd
import pyperclip

time.sleep(2)
df = pd.read_csv(r"/Users/danielbarbosadeoliveira/Downloads/contacts.csv", sep=";")
df = df.dropna(axis=1, how="all")

pyautogui.hotkey("ctrl", "pgup")
```

```python
for email in df["E-mail 1 - Value"]:
    time.sleep(2)
    # clicar no botao escrever
    escrever = pyautogui.locateCenterOnScreen("escrever.png")
    pyautogui.click(escrever)
    time.sleep(1)
    
    # escrever o email e enter
    pyautogui.write("dan.dam@hotmail.com")
    pyautogui.press("enter")
    time.sleep(1)
    # escrever titulo do email
    pyautogui.press("tab")
    pyautogui.write("Daniel Caloteiro")
    
    # corpo do email
    pyautogui.press("tab")
    
    texto = """
    Coe Lira,
    
    Para de dar calote na Hashtag e paga as parcelas ai, namoral.
    
    Abs e tmj"""
    pyperclip.copy(texto)
    pyautogui.hotkey("command", "v")
    time.sleep(0.5)
    pyautogui.hotkey("command", "enter")
```

```python
time.sleep(2)
escrever = pyautogui.locateCenterOnScreen("escrever.png")
pyautogui.click(escrever)
time.sleep(1)

# escrever o email e enter
pyautogui.write(df["E-mail 1 - Value"][0])
pyautogui.press("enter")
```
