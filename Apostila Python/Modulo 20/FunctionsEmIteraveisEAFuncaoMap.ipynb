* Exempo: Digamos que eu tenha uma function que corrige um código de um produto (semelhante ao que fizemos na sessão de function do curso)

```python
def padronizar_texto(texto):
    texto = texto.casefold()
    texto = texto.replace(" ", "")
    texto = texto.strip()
    return texto
```

```python
produtos = [' ABC12 ', 'abc34', 'AbC37', 'beb12', ' B5A151', 'BEB23']
```

```python
produtos = list(map(padronizar_texto, produtos))
print(produtos)
```
