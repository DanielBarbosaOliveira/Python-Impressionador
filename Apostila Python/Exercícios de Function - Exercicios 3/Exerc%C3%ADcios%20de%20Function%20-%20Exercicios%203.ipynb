## 1. Cálculo do Percentual e da Lista de Vendedores

- Queremos criar uma function que consiga identificar os vendedores que bateram uma meta, mas além disso, consigo já me dar como resposta o cálculo do % da lista de vendedores que bateu a meta (para eu não precisar calcular manualmente depois)
- Essa function deve receber 2 informações como parâmetro: a meta e um dicionário com os vendedores e suas vendas. E me dar 2 respostas: uma lista com o nome dos vendedores que bateram a meta e o % de vendedores que bateu a meta.

```python
meta = 10000
vendas = {
    'João': 15000,
    'Julia': 27000,
    'Marcus': 9900,
    'Maria': 3750,
    'Ana': 10300,
    'Alon': 7870,
}
```

```python
#crie sua function aqui
def calculo_de_vededores_baterm_meta_retorno_percentual(valor_meta, dicionario_vendas):
    bateram_meta = []
    for nome in dicionario_vendas:
        if dicionario_vendas[nome] >= valor_meta:
            bateram_meta.append(nome)
    percentual_bateram_meta = len(bateram_meta) / len(dicionario_vendas)
    return bateram_meta, percentual_bateram_meta
```

```python
#aplique sua function nas informações para ver se está funcionando
vendedores_bateram_meta, percentual_bateram_meta = calculo_de_vededores_baterm_meta_retorno_percentual(meta, vendas)
print("vendedores que bateram a meta:")
for vendedor in vendedores_bateram_meta:
    print(vendedor)
print(f'Representando {percentual_bateram_meta:.2%} de todos os vendedores!')
```
