```python
import time
```

* Marco Zero: EPOCH (01 de Janeiro de 1970 as 00:00:00)
* time() retorna quantos segundos se passaram desde EPOCH

```python
print(time.time())
```

Isso é muito útil para medir quanto tempo que um código leva para rodar

```python
tempo_inicial = time.time()

for i in range(100000000):
    pass

tempo_final = time.time()

print(f"O meu programa demorou {tempo_final - tempo_inicial} segundos.")
```

* ctime() retorna a data em string

```python
data_hoje = time.ctime()
print(data_hoje)
```

Fazer o programa esperar alguns segundos (muito útil quando temos que esperar um programa ou uma página carregar)

```python
time.sleep()
```

Pegar informações de dias, hora, minuto, segundo, tudo detalhado:
* gmtime()
* localtime()

```python
data_global = time.gmtime()
print(data_global)
data_local = time.localtime()
print(data_local)
```

Qualquer struct_time permite você pegar as informações da data fazendo:
* gmtime().parametro
* localtime().parametro

```python
dia_hoje = data_local.tm_mday
mes_hoje = data_local.tm_mon

print(f"Hoje é dia {dia_hoje}")
print(f"Estamos no mês {mes_hoje}")
```

Agora, para trabalhar com formatações de datas, recomendo fortemente você ter esse link: https://docs.python.org/3/library/time.html#time.strftime

E para formatar datas para serem exibidas em uma mensagem do jeito que a gente quer ?

```python
data_text = time.strftime("%d/%m/%Y %A", data_local)
print(data_text)
```

E se eu quisero texto em Portugues (do Mês e dia da semana). Ai você precisa dizer para o Python o seu 'local', usando a biblioteca locale. link:
http://docs.python.org/3/library/locale.html

```python
import locale

locale.setlocale(locale.LC_TIME, 'pt_BR')

data_texto = time.strftime("%d/%m/%Y %A %B")

print(data_texto)
```

E para ler arquivos com datas formatadas de maneira diferente ?

```python
lista_datas = ["01 de Junho de 2026", "01 de Janeiro de 2027", "15 de Fevereiro de 2027", "18 de Dezembro de 2027"]

for data in lista_datas:
    data_estruturada = time.strptime(data, '%d de %B de %Y')
    print(data_estruturada)
```
