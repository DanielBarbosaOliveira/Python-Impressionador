# Automação de Sistemas e Processos com Python

### Desafio:

Para controle de custos, todos os dias, seu chefe pede um relatório com todas as compras de mercadorias da empresa.
O seu trabalho, como analista, é enviar um e-mail para ele, assim que começar a trabalhar, com o total gasto, a quantidade de produtos compradas e o preço médio dos produtos.

E-mail do seu chefe: para o nosso exercício, coloque um e-mail seu como sendo o e-mail do seu chefe<br>
Link de acesso ao sistema da empresa: https://pages.hashtagtreinamentos.com/aula1-intensivao-sistema

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado

```python
import pyperclip
import pyautogui
import time

pyautogui.PAUSE = 1
# Passo 1: Entrar no sistema da empresa (no link)
time.sleep(2)
pyautogui.hotkey("ctrl", "t")
pyautogui.write("https://pages.hashtagtreinamentos.com/aula1-intensivao-sistema")
pyautogui.press("enter")

# Aguardar talvez o tempo do site abrir
time.sleep(3)

# Passo 2: Fazer login
# Clicar no espaço do login
pyautogui.click(x=3509, y=485)
# Escrever usuário
pyautogui.write("meu_login")

# Escrever senha
pyautogui.press('tab')
pyautogui.write("minha_senha")

# Clicar em acessar
pyautogui.click(x=3565, y=661)

time.sleep(3)

# Passo 3: Exportar a base de dados
pyautogui.rightClick(x=2954, y=480) # Clicar com o botão direito na base de dados
pyautogui.press("down", presses=2) # Descer a seleção até o botão download
pyautogui.press('enter') # Baixar arquivo
time.sleep(3)
```

```python
# Passo 4: Calcular os indicadores
import pandas as pd

tabela = pd.read_csv(r"C:\Users\santo\Downloads\Compras.csv", sep=';')
display(tabela)
total_gasto = tabela["ValorFinal"].sum()
quantidade = tabela["Quantidade"].sum()
preco_medio = total_gasto / quantidade
print(f"{total_gasto:,.2f}", quantidade, preco_medio, sep='\n')
```

```python

pyautogui.PAUSE=1.5
# Passo 5: Enviar um e-mail para o meu chefe
# Entrar no gmail
pyautogui.hotkey("ctrl", "t")
pyautogui.write("gmail.com")
pyautogui.press("enter")
#aguardar gmail carregar
time.sleep(3)
# Clicar em escrever
pyautogui.click(x=2510, y=255)
# Escrever o e-mail do remetente
time.sleep(2)
pyautogui.write('dan.dam@hotmail.com')
pyautogui.press('enter')

pyautogui.press('tab')
pyperclip.copy('Relatório de Vendas')
pyautogui.hotkey("ctrl", "v")
pyautogui.press('tab')
corpo_email = f"""
Prezados,
Segue o relatório de compras

Total Gasto: R$ {total_gasto:,.2f}
Quantidade de Produtos: {quantidade} unidades
Preço Médio: R$ {preco_medio:,.2f}

Qualquer dúvida, é só falar.
Att., Daniel
"""
pyautogui.write("Corpo do e-mail")
pyautogui.hotkey('ctrl', 'enter')
```
